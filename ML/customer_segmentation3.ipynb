```python
# Exercise 3: Self-Reflection Prompt for Improving Output

# -----------------------------------
# ORIGINAL SUMMARY
# -----------------------------------

original_summary = """
Artificial intelligence helps businesses work faster and make better decisions.
Companies can use AI in many departments, including customer service, marketing,
and operations. AI is becoming more common because it can automate tasks and
help companies analyze information.
"""

print("ORIGINAL SUMMARY:")
print(original_summary)


# -----------------------------------
# SELF-REFLECTION / CRITIQUE
# -----------------------------------

print("\nSELF-REFLECTION / CRITIQUE:")

critique = [
    "Accuracy: The summary is generally accurate, but it is very broad.",
    "Clarity: The wording is understandable, but phrases such as 'work faster' are vague.",
    "Audience: The language is appropriate for a college business student.",
    "Length: The summary is shorter than the required 80-120 words.",
    "Focus: It mentions business areas but does not provide enough specific examples of AI creating value.",
    "Formatting: The summary should remain one clear paragraph."
]

for item in critique:
    print("-", item)


# -----------------------------------
# IMPROVED SUMMARY
# -----------------------------------

improved_summary = """
Artificial intelligence helps businesses improve efficiency, analyze data,
and support better decision-making. For example, companies can use AI-powered
chatbots in customer service to answer common questions quickly and reduce
employee workload. Marketing teams can use AI to analyze customer behavior and
create more personalized campaigns. Businesses can also use AI in operations
to forecast demand, manage inventory, and identify patterns in large datasets.
These applications can save time and improve productivity, but companies
should still use human oversight to review important decisions and ensure
that AI systems are used responsibly.
"""

print("\nIMPROVED SUMMARY:")
print(improved_summary)
```
