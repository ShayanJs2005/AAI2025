```python
# Exercise 2: Code Generation with ReACT Prompting

# -----------------------------------
# PLAN
# -----------------------------------

print("PLAN:")
print("1. Store the customer satisfaction scores.")
print("2. Calculate the average score.")
print("3. Compare the average with the target.")
print("4. Handle an empty list to avoid division by zero.")
print("5. Print a recommendation.")

# -----------------------------------
# INPUT
# -----------------------------------

scores = [4, 5, 3, 4, 2]
target = 4.0

# -----------------------------------
# ACTION - Generate and run code
# -----------------------------------

print("\nACTION:")
print("Running customer satisfaction analysis...")

if len(scores) == 0:
    average_score = None
else:
    average_score = sum(scores) / len(scores)

# -----------------------------------
# OBSERVATION
# -----------------------------------

print("\nOBSERVATION:")

if average_score is None:
    print("No scores were provided.")
else:
    print(f"Average Score: {average_score:.2f}")
    print(f"Target Score: {target:.2f}")

# -----------------------------------
# FINAL RESULT
# -----------------------------------

print("\nFINAL RESULT:")

if average_score is None:
    print("Recommendation: Add customer satisfaction data before analyzing results.")

elif average_score >= target:
    print("Target Met: Yes")
    print("Recommendation: Continue the current customer service approach.")

else:
    print("Target Met: No")
    print("Recommendation: Review customer feedback and improve the service process.")
```

Exercise 2: Code Generation with ReACT Prompting

Goal:
Generate Python code using a ReACT-style process that includes planning, execution, observation, correction, and a final result.

Tools Used:
- Google Colab
- Python and Claude with Chat


ReACT Process:
1. Plan:
The AI first identifies what the program needs to do.

2. Action:
The AI generates Python code to calculate the average customer satisfaction score and compare it with a target.

3. Run:
The code is executed in Google Colab.

4. Observation:
The output is reviewed to check whether the result is correct and whether the code handles possible errors.

5. Fix:
If a problem is found, the code is revised.

6. Final Result:
The corrected code is executed again and the final output is displayed.

Testing and Iteration:

First Version:
The first version used:

average_score = sum(scores) / len(scores)

Problem:
This works when the list contains values, but it would cause a division-by-zero error if the list were empty.

Observation:
The code needed basic error handling so it would not crash when no customer satisfaction scores were provided.

Fix:
I added:

if len(scores) == 0:
    average_score = None
else:
    average_score = sum(scores) / len(scores)

Result:
The revised code now works with the normal sample data and also handles an empty list without producing an error.

Why This Follows ReACT:
The process shows a clear cycle of planning the task, generating and running code, observing a weakness, fixing the code, and running the improved version.
