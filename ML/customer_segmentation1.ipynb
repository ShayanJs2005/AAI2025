```python
# Exercise 1: Prompt Chaining for a Customer Support AI

# Customer message
customer_message = "My package says delivered, but I never received it."

# -------------------------
# Step 1: Classify the issue
# -------------------------

def classify_issue(message):
    message = message.lower()

    if "package" in message or "delivered" in message or "shipping" in message:
        return "Shipping"
    elif "refund" in message or "charged" in message or "payment" in message:
        return "Billing"
    elif "login" in message or "password" in message or "error" in message:
        return "Technical Support"
    else:
        return "General Inquiry"

issue_category = classify_issue(customer_message)

print("STEP 1 - ISSUE CLASSIFICATION")
print("Customer Message:", customer_message)
print("Category:", issue_category)


# --------------------------------
# Step 2: Identify missing info
# Uses output from Step 1
# --------------------------------

def identify_missing_info(category):
    if category == "Shipping":
        return "Order number and confirmation of delivery address"
    elif category == "Billing":
        return "Order number and description of the billing issue"
    elif category == "Technical Support":
        return "Device type and description of the error"
    else:
        return "More details about the customer's request"

missing_info = identify_missing_info(issue_category)

print("\nSTEP 2 - MISSING INFORMATION")
print("Based on category:", issue_category)
print("Information needed:", missing_info)


# --------------------------------
# Step 3: Generate response
# Uses output from Steps 1 and 2
# --------------------------------

def generate_response(category, needed_info):
    if category == "Shipping":
        return (
            "I'm sorry your package has not arrived even though it shows as delivered. "
            f"Please provide your {needed_info.lower()}. "
            "Once this information is verified, we can review the delivery status "
            "and determine the appropriate next step."
        )

    elif category == "Billing":
        return (
            f"Please provide your {needed_info.lower()} so we can review the issue."
        )

    elif category == "Technical Support":
        return (
            f"Please provide your {needed_info.lower()} so we can troubleshoot the problem."
        )

    else:
        return "Please provide more information so we can assist you."

response = generate_response(issue_category, missing_info)

print("\nSTEP 3 - CUSTOMER RESPONSE")
print(response)


# --------------------------------
# Step 4: Determine escalation
# Uses output from previous steps
# --------------------------------

def escalation_decision(category, message):
    message = message.lower()

    if "theft" in message or "fraud" in message:
        return "Escalate: Yes - Possible theft or fraud requires human review."

    elif category == "Shipping":
        return (
            "Escalate: No - Standard delivery verification should be completed first."
        )

    else:
        return "Escalate: No - The issue can continue through normal support."

escalation = escalation_decision(issue_category, customer_message)

print("\nSTEP 4 - ESCALATION DECISION")
print(escalation)
```
